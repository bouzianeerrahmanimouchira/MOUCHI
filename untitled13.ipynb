### Étape : Nettoyage et Préparation des Données
Nous allons identifier les valeurs manquantes, supprimer les doublons et standardiser les chaînes de caractères (en enlevant les espaces superflus et en uniformisant la casse).

```python
import pandas as pd
import numpy as np

# 1. Vérifier si df est bien chargé, sinon le recharger
if 'df' not in locals() and 'df' not in globals():
    file_path = '/content/antibiotics_data/Antibiotics Resistance pattern and Resistance gene/Raw data.xlsx'
    df = pd.read_excel(file_path)

print("--- Données avant nettoyage ---")
print(f"Nombre de lignes : {df.shape[0]}")
print(f"Nombre de colonnes : {df.shape[1]}")

# 2. Vérification des doublons
doublons = df.duplicated().sum()
print(f"\nNombre de lignes doublons détectées : {doublons}")
if doublons > 0:
    df = df.drop_duplicates()
    print("Doublons supprimés !")

# 3. Analyse des valeurs manquantes
valeurs_manquantes = df.isnull().sum()
print("\nValeurs manquantes par colonne (top 15) :")
print(valeurs_manquantes[valeurs_manquantes > 0].head(15))

# Remplissage ou traitement des valeurs manquantes (ici exemple de nettoyage simple)
# Pour l'âge, on peut imputer par la médiane si nécessaire
if 'AGE' in df.columns and df['AGE'].isnull().any():
    med_age = df['AGE'].median()
    df['AGE'] = df['AGE'].fillna(med_age)
    print(f"\nValeurs manquantes dans 'AGE' remplacées par la médiane : {med_age}")
```

```python
# 4. Standardisation des variables textuelles/catégoriques
# (Mise en majuscules, suppression des espaces blancs en début/fin)
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

for col in categorical_cols:
    # Conversion en chaîne de caractères, mise en majuscules et suppression des espaces
    df[col] = df[col].astype(str).str.strip().str.upper()
    # Remplacer les chaînes vides ou 'NAN'/'NAT' par des valeurs manquantes réelles
    df[col] = df[col].replace({'NAN': np.nan, 'NAT': np.nan, 'NONE': np.nan, '': np.nan})

print("Standardisation des colonnes textuelles terminée.")
display(df.head())
```

### Étape : Division des données (75% Entraînement / 25% Évaluation)
Nous allons utiliser `train_test_split` de `scikit-learn`. Pour cet exemple, nous allons essayer de prédire la résistance à un antibiotique spécifique (par exemple, `IPM` - Imipénem) en utilisant l'âge, le genre et l'isolat bactérien comme caractéristiques ($X$).

```python
from sklearn.model_selection import train_test_split

# Sélection des caractéristiques (X) et de la cible (Y)
# Ici, nous utilisons l'AGE, le GENDER et l'ISOLATES pour prédire la résistance à l'IPM
features = ['AGE', 'GENDER', 'ISOLATES']
target = 'IPM'

# Filtrer le dataset pour enlever les lignes où la cible (IPM) est manquante (NaN)
df_ml = df.dropna(subset=[target]).copy()

X = df_ml[features]
y = df_ml[target]

# Division des données : 75% entraînement et 25% évaluation
X_train, X_eval, y_train, y_eval = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y if y.nunique() > 1 else None # Stratification si possible pour conserver les proportions
)

print("--- Division des données complétée ---")
print(f"Taille de l'ensemble d'entraînement (75%) : {X_train.shape[0]} échantillons")
print(f"Taille de l'ensemble d'évaluation (25%)  : {X_eval.shape[0]} échantillons")
```

```python
# Aperçu rapide des ensembles d'entraînement
print("\nAperçu des caractéristiques d'entraînement (X_train) :")
display(X_train.head())

print("\nDistribution de la cible dans l'entraînement (y_train) :")
print(y_train.value_counts(dropna=False))
```

### Étape : Encodage, Entraînement et Évaluation du Modèle de Machine Learning
Nous allons :
1. Utiliser un pipeline de prétraitement pour encoder les variables catégoriques (`GENDER`, `ISOLATES`) et normaliser l'âge.
2. Entraîner un modèle d'Arbre de Décision (`DecisionTreeClassifier`) sur nos 65 lignes d'entraînement.
3. Prédire et évaluer l'efficacité du modèle sur les 22 lignes de test à l'aide d'un rapport de classification.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score

# 1. Définition du prétraitement pour les colonnes numériques et catégoriques
numerical_features = ['AGE']
categorical_features = ['GENDER', 'ISOLATES']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)

# 2. Création du pipeline associant le prétraitement et le classifieur
model_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', DecisionTreeClassifier(random_state=42, max_depth=5))
])

# Encodage de la variable cible y ('S' et 'R' en valeurs numériques)
y_train_encoded = y_train.map({'S': 0, 'R': 1})
y_eval_encoded = y_eval.map({'S': 0, 'R': 1})

# 3. Entraînement du modèle
model_pipeline.fit(X_train, y_train_encoded)
print("Modèle entraîné avec succès !")
```

```python
# 4. Évaluation du modèle sur l'ensemble d'évaluation (25%)
y_pred = model_pipeline.predict(X_eval)

accuracy = accuracy_score(y_eval_encoded, y_pred)
print(f"Précision globale (Accuracy) : {accuracy * 100:.2f}%")

print("\nRapport de classification complet :")
print(classification_report(y_eval_encoded, y_pred, target_names=['Sensible (S)', 'Résistant (R)'], zero_division=0))
```

### Étape : Évaluation Statistique Approfondie selon les Normes Standard
Nous allons calculer et afficher les métriques statistiques requises pour valider l'efficacité biologique de l'algorithme : matrice de confusion, sensibilité (recall), spécificité et score AUC.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, roc_curve, roc_auc_score, precision_recall_fscore_support

# 1. Calcul de la Matrice de Confusion
cm = confusion_matrix(y_eval_encoded, y_pred)

# Extraction des True Negatives (TN), False Positives (FP), False Negatives (FN), True Positives (TP)
tn, fp, fn, tp = cm.ravel()

# 2. Calcul des métriques cliniques clés
sensibilite = tp / (tp + fn) if (tp + fn) > 0 else 0  # Rappel de la classe Résistant
specificite = tn / (tn + fp) if (tn + fp) > 0 else 0  # Rappel de la classe Sensible

print("--- Évaluation Statistique Clinique ---")
print(f"Vrais Sensibles bien identifiés (TN)   : {tn}")
print(f"Faux Résistants détectés à tort (FP)   : {fp}")
print(f"Faux Sensibles (Résistances manquées) (FN) : {fn}")
print(f"Vrais Résistants bien identifiés (TP)  : {tp}")
print(f"\nSensibilité (Taux de détection de la résistance) : {sensibilite * 100:.2f}%")
print(f"Spécificité (Taux de détection de la sensibilité) : {specificite * 100:.2f}%")
```

```python
# 3. Visualisation de la Matrice de Confusion
fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Sensible (S)', 'Résistant (R)'],
            yticklabels=['Sensible (S)', 'Résistant (R)'], ax=ax)
plt.title("Matrice de Confusion de la Résistance à l'IPM")
plt.ylabel('Classe Réelle')
plt.xlabel('Classe Prédite')
plt.tight_layout()
plt.show()
```

```python
# 4. Courbe ROC (Receiver Operating Characteristic) et score AUC
# On récupère les probabilités de prédiction pour la classe positive (Résistant)
y_probs = model_pipeline.predict_proba(X_eval)[:, 1]

try:
    auc_score = roc_auc_score(y_eval_encoded, y_probs)
    fpr, tpr, thresholds = roc_curve(y_eval_encoded, y_probs)

    fig2 = plt.figure(figsize=(7, 5))
    plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'Courbe ROC (AUC = {auc_score:.2f})')
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel("Taux de Faux Positifs (1 - Spécificité)")
    plt.ylabel("Taux de Vrais Positifs (Sensibilité)")
    plt.title("Courbe ROC de l'évaluation du modèle")
    plt.legend(loc="lower right")
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()
except Exception as e:
    print("Impossible de tracer la courbe ROC car une seule classe est présente ou une erreur est survenue:", e)
```

### Étape : Discussion Statistique de l'Éfficacité de Prédiction (Comparaison Réel vs Prédit en %)
Ce tableau croise les caractéristiques réelles des patients ($X$), leur statut de résistance réel ($Y$ Réel) et la prédiction faite par l'arbre de décision ($Y$ Prédit) afin d'illustrer visuellement les performances cliniques.

```python
import pandas as pd
import numpy as np
import os
import zipfile
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier

# Définition des chemins d'accès
file_path = '/content/antibiotics_data/Antibiotics Resistance pattern and Resistance gene/Raw data.xlsx'
zip_path = '/content/Antibiotics Resistance pattern and Resistance gene(1).zip'
extract_path = '/content/antibiotics_data'

# Sécurité : Si le fichier Excel est manquant, on tente de ré-extraire l'archive zip d'abord
if not os.path.exists(file_path):
    if os.path.exists(zip_path):
        print("Le dossier extrait est absent. Ré-extraction automatique du fichier zip en cours...")
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_path)
        print("Extraction complétée.")
    else:
        raise FileNotFoundError(f"Le fichier Excel et l'archive zip {zip_path} sont tous deux introuvables. Veuillez importer à nouveau l'archive.")

# Sécurité : Si les variables de l'évaluation ne sont pas définies, on recrée l'ensemble de données et le modèle
if 'X_eval' not in locals() or 'y_eval_encoded' not in locals() or 'y_pred' not in locals():
    print("Variables de modélisation manquantes détectées en mémoire. Rechargement et réentraînement automatique...")

    # 1. Chargement et nettoyage rapide
    df_temp = pd.read_excel(file_path)
    if 'AGE' in df_temp.columns and df_temp['AGE'].isnull().any():
        df_temp['AGE'] = df_temp['AGE'].fillna(df_temp['AGE'].median())

    categorical_cols = df_temp.select_dtypes(include=['object', 'category']).columns.tolist()
    for col in categorical_cols:
        df_temp[col] = df_temp[col].astype(str).str.strip().str.upper()
        df_temp[col] = df_temp[col].replace({'NAN': np.nan, 'NAT': np.nan, 'NONE': np.nan, '': np.nan})

    df_ml = df_temp.dropna(subset=['IPM']).copy()
    X = df_ml[['AGE', 'GENDER', 'ISOLATES']]
    y = df_ml['IPM']

    # 2. Division
    X_train, X_eval, y_train, y_eval = train_test_split(
        X, y, test_size=0.25, random_state=42, stratify=y if y.nunique() > 1 else None
    )

    # 3. Pipeline & Entraînement
    preprocessor = ColumnTransformer(
        transformers=[
            ('num', StandardScaler(), ['AGE']),
            ('cat', OneHotEncoder(handle_unknown='ignore'), ['GENDER', 'ISOLATES'])
        ]
    )
    model_pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', DecisionTreeClassifier(random_state=42, max_depth=5))
    ])

    y_train_encoded = y_train.map({'S': 0, 'R': 1})
    y_eval_encoded = y_eval.map({'S': 0, 'R': 1})

    model_pipeline.fit(X_train, y_train_encoded)
    y_pred = model_pipeline.predict(X_eval)

# 1. Préparation du tableau de confrontation Réel vs Prédit
df_discussion = X_eval.copy()

# Associer les classes textuelles d'origine pour une lecture claire
map_classes = {0: 'Sensible (S)', 1: 'Résistant (R)'}

df_discussion['Y_Réel'] = y_eval_encoded.map(map_classes)
df_discussion['Y_Prédit'] = pd.Series(y_pred, index=X_eval.index).map(map_classes)

# Déterminer si la prédiction est correcte
df_discussion['Résultat_Prédiction'] = np.where(
    df_discussion['Y_Réel'] == df_discussion['Y_Prédit'],
    'SUCCÈS',
    'ÉCHEC'
)

# 2. Calcul des pourcentages globaux d'efficacité
total_echantillons = len(df_discussion)
succes_count = (df_discussion['Résultat_Prédiction'] == 'SUCCÈS').sum()
echec_count = (df_discussion['Résultat_Prédiction'] == 'ÉCHEC').sum()

pct_succes = (succes_count / total_echantillons) * 100
pct_echec = (echec_count / total_echantillons) * 100

print("=== RAPPORT STATISTIQUE D'ÉFFICACITÉ ===")
print(f"Nombre total de cas testés : {total_echantillons}")
print(f"Prédictions Correctes (Succès) : {succes_count} ({pct_succes:.2f}%)")
print(f"Prédictions Incorrectes (Échecs) : {echec_count} ({pct_echec:.2f}%)")
print("========================================\n")

# Affichage propre du tableau de discussion
display(df_discussion)
```
